#### Use WordNet to identify semantic relationships such as synonymy, antonymy, and hypernymy in text data

```python
import nltk
from nltk.corpus import wordnet as wn

# Download required datasets
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
def get_semantic_relations(word):
    synonyms = set()
    antonyms = set()
    hypernyms = set()

    for syn in wn.synsets(word):
        for lemma in syn.lemmas():
            synonyms.add(lemma.name())

            if lemma.antonyms():
                antonyms.add(lemma.antonyms()[0].name())

        for hyper in syn.hypernyms():
            hypernyms.add(hyper.name().split('.')[0])

    return {
        'word': word,
        'synonyms': list(synonyms),
        'antonyms': list(antonyms),
        'hypernyms': list(hypernyms),
    }
```

```python
# Example usage
words = ['good', 'dog', 'car']

for w in words:
    result = get_semantic_relations(w)
    print('\nWord:', result['word'])
    print('Synonyms:', result['synonyms'][:5])
    print('Antonyms:', result['antonyms'])
    print('Hypernyms:', result['hypernyms'][:5])
```
